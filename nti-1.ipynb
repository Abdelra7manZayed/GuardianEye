```python
cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    print("Error: Cannot open video")
else:
    print("Video opened successfully")

fps = int(cap.get(cv2.CAP_PROP_FPS))

print("FPS:", fps)
```

```python
ret, frame = cap.read()

print(ret)
print(frame.shape)
```

```python
import cv2
import numpy as np

# Violation score
violation_score = 0
threshold = 35

unsafe_classes = [
    "no_helmet",
    "no_vest",
    "no_vest_no_helmet"
]

while True:

    ret, frame = cap.read()

    if not ret:
        break

    # YOLO Detection
    results = model(frame, conf=0.5)

    unsafe = False
    unsafe_boxes = []  # store boxes of violators only

    for r in results:
        for box in r.boxes:
            cls_id = int(box.cls[0])
            cls_name = model.names[cls_id]

            if cls_name in unsafe_classes:
                unsafe = True
                unsafe_boxes.append(box)

    # Update violation score
    if unsafe:
        violation_score += 1
    else:
        violation_score = max(0, violation_score - 1)

    # Build the frame to display
    display_frame = frame.copy()

    if unsafe:
        # Light blur on the whole frame (kernel size controls strength;
        # smaller odd numbers = lighter blur, e.g. (9,9) or (15,15))
        display_frame = cv2.GaussianBlur(display_frame, (15, 15), 0)

        # Draw a clear bounding box + label around each violator,
        # ON TOP of the blur, so the box itself stays sharp
        for box in unsafe_boxes:
            x1, y1, x2, y2 = map(int, box.xyxy[0])
            cls_id = int(box.cls[0])
            cls_name = model.names[cls_id]
            conf = float(box.conf[0])

            # Optionally: un-blur just the region inside the box so the
            # violator himself stays sharp while the background is blurred
            display_frame[y1:y2, x1:x2] = frame[y1:y2, x1:x2]

            cv2.rectangle(display_frame, (x1, y1), (x2, y2), (0, 0, 255), 2)
            label = f"{cls_name} {conf:.2f}"
            cv2.putText(
                display_frame, label, (x1, max(y1 - 10, 0)),
                cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 0, 255), 2
            )

    # Alert
    if violation_score >= threshold:
        cv2.imwrite("alert_frame.jpg", display_frame)
        print("🚨 PPE Violation Detected")
        violation_score = 0

    # Display video
    cv2.imshow("PPE Detection", display_frame)

    # Press q to exit
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
